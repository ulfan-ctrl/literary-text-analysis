## Step 2: Word Frequency Analysis

In this step, I will count the words in each story and identify the most frequently used words.

```python
print ("Story 1 ending:")
print (story1 [-500:])
```

```python
# Count the words in both stories

words1 = story1.lower().split()

words2 = story2.lower().split()

print("The Gift of the Magi:", len(words1), "words")

print("The Story of an Hour:", len(words2), "words")
```

```python
# Remove punctuation and prepare words


import re

story1_clean = story1.lower().replace("’", "'")

story2_clean = story2.lower().replace("’", "'")

clean_words1 = re.findall(r"[a-z]+(?:'[a-z]+)*", story1_clean)

clean_words2 = re.findall(r"[a-z]+(?:'[a-z]+)*", story2_clean)

print("Story 1:", len(clean_words1), "words")

print("Story 2:", len(clean_words2), "words")
```

```python
# Count unique words in both stories

unique_words1 = set(clean_words1)

unique_words2 = set(clean_words2)

print("The Gift of the Magi:", len(unique_words1), "unique words")

print("The Story of an Hour:", len(unique_words2), "unique words")
```

```python
# Find the 10 most frequent words

from collections import Counter

frequency1 = Counter(clean_words1)

frequency2 = Counter(clean_words2)

print("The Gift of the Magi:")

print(frequency1.most_common(10))

print("\nThe Story of an Hour:")

print(frequency2.most_common(10))
```

```python
# Remove common stopwords

from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

filtered_words1 = [

    word for word in clean_words1

        if word not in ENGLISH_STOP_WORDS and len (word)>1

        ]
filtered_words2 = [

            word for word in clean_words2

                if word not in ENGLISH_STOP_WORDS and len (word)>1

                ]
print("The Gift of the Magi:")
print(Counter(filtered_words1).most_common(10))
print("\nThe Story of an Hour:")
print(Counter(filtered_words2).most_common(10))
```

## Step 3: Data Visualization

In this step, I will create bar charts to visualize the 10 most frequent words in each story after removing common stopwords and single-letter words.

```python
# Visualize the most frequent words in the first story

import matplotlib.pyplot as plt

top_words1 = Counter(filtered_words1).most_common(10)

words = [item[0] for item in top_words1]

counts = [item[1] for item in top_words1]

plt.bar(words, counts)

plt.title("The Gift of the Magi - Top 10 Words")

plt.xlabel("Words")

plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.show()
```

```python
# Visualize the most frequent words in the second story

top_words2 = Counter(filtered_words2).most_common(10)

words2 = [item[0] for item in top_words2]

counts2 = [item[1] for item in top_words2]

plt.bar(words2, counts2)

plt.title("The Story of an Hour - Top 10 Words")

plt.xlabel("Words")

plt.ylabel("Frequency")

plt.xticks(rotation=45)

plt.show()
```

## Step 4: Comparing Word Frequencies

The two stories have different lengths. To make a fair comparison, I will calculate the relative frequency of words per 1,000 words.

```python
# Calculate relative frequency per 1000 words

word = "hair"

count1 = clean_words1.count(word)

count2 = clean_words2.count(word)

relative1 = count1 / len(clean_words1) * 1000

relative2 = count2 / len(clean_words2) * 1000

print("Word:", word)

print("The Gift of the Magi:", round(relative1, 2))

print("The Story of an Hour:", round(relative2, 2))
```

```python
# Compare selected words in both stories

selected_words = ["hair", "love", "life", "free", "door"]

for word in selected_words:

    count1 = clean_words1.count(word)

count2 = clean_words2.count(word)

relative1 = count1 / len(clean_words1) * 1000

relative2 = count2 / len(clean_words2) * 1000

print(word)
print("Story 1:", round(relative1, 2))
print("Story 2:", round(relative2, 2))
print()
```

```python
# Check the exact word frequencies

for word in selected_words:
              count1 = clean_words1.count(word)

              count2 = clean_words2.count(word)
              print(word, ":", count1, "in Story 1,", count2, "in Story 2")
```

```python
# Compare relative word frequencies using a bar chart

import matplotlib.pyplot as plt

import numpy as np

selected_words = ["hair", "love", "life", "free", "door"]

values1 = []

values2 = []

for word in selected_words:

    values1.append(clean_words1.count(word) / len(clean_words1) * 1000)

    values2.append(clean_words2.count(word) / len(clean_words2) * 1000)

x = np.arange(len(selected_words))

width = 0.35

plt.bar(x - width/2, values1, width, label="The Gift of the Magi")

plt.bar(x + width/2, values2, width, label="The Story of an Hour")

plt.xticks(x, selected_words)

plt.xlabel("Words")

plt.ylabel("Frequency per 1,000 words")

plt.title("Comparison of Word Frequencies")

plt.legend()

plt.show()
```

### Step 5: Results and Discussion

The analysis shows some differences in vocabulary between the two short stories.

In The Gift of the Magi, the word “hair” appears frequently, while it does not appear in The Story of an Hour.

In The Story of an Hour, words such as “free”, “life”, and “door” have higher relative frequencies.

These differences may reflect the different themes of the stories. The Gift of the Magi focuses on love and sacrifice, while The Story of an Hour explores freedom and personal independence. For example, in "The Gift of the Magi", Della’s hair plays an important role because she sells it to buy a Christmas present for Jim. This helps explain why the word “hair” appears frequently in the story. In "The Story of an Hour", the word “free” is connected to Louise Mallard’s feelings about her independence. These examples suggest that word-frequency analysis can help identify important elements of a story, but their meaning becomes clearer through close reading.

Since the stories have different lengths, I compared word frequencies per 1,000 words to make the comparison fairer. For example, the word “hair” appears 6.76 times per 1,000 words in "The Gift of the Magi", but it does not appear in "The Story of an Hour". In contrast, the word “free” appears 4.92 times per 1,000 words in "The Story of an Hour" and does not appear in "The Gift of the Magi". The word “door” also has a higher relative frequency in The Story of an Hour (5.91) than in The Gift of the Magi (1.93). These numerical results are consistent with the patterns shown in the comparison chart.

However, word frequency alone cannot fully explain the meaning of a literary text. The analysis focuses on individual words and does not examine how they are used in sentences. Also, the results are based on only two short stories. Therefore, the patterns should be interpreted together with a close reading of the texts.
